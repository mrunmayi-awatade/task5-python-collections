# 1. List creation and basic operations

```python
students = ["Mrunmayi", "Rahul", "Priya", "Aman"]
print("Student List:", students)
print("First Student:", students[0])
print("Number of Students:", len(students))
```

```python
#List methods
students = ["Mrunmayi", "Rahul", "Priya", "Aman"]
students.append("Neha")
students.remove("Rahul")
students.sort()
print("Updated List:", students)
print("Is Priya present?", "Priya" in students)
```

```python
#Tuple creation and operations
student_marks = (75, 82, 68, 90, 76)
print("Tuple:", student_marks)
print("First Mark:", student_marks[0])
print("Number of Marks:", len(student_marks))
print("Highest Mark:", max(student_marks))
```

```python
#Tuple methods
marks = (75, 82, 68, 82, 90, 82)
print("Number of 82:", marks.count(82))
print("Position of 90:", marks.index(90))
```

```python
#Set creation and unique values
marks = {75, 82, 68, 90, 75, 82}
print("Set:", marks)
print("Number of Unique Marks:", len(marks))
print("Is 90 present?", 90 in marks)
```

```python
#Set operations
set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}
print("Union:", set_a | set_b)
print("Intersection:", set_a & set_b)
print("Difference:", set_a - set_b)
```

```python
#Dictionary creation and accessing values
student = {
    "name": "Mrunmayi",
    "age": 20,
    "branch": "CSE-DS",
    "marks": 85
}
print("Student Details:", student)
print("Name:", student["name"])
print("Marks:", student["marks"])
```

```python
#Dictionary methods
student = {
    "name": "Mrunmayi",
    "age": 20,
    "branch": "CSE-DS",
    "marks": 85
}
student["city"] = "Vijayawada"
student["marks"] = 90
print("Updated Dictionary:", student)
print("Keys:", student.keys())
print("Values:", student.values())
student.pop("age")
print("After Removing Age:", student)
```

```python
#Using different collections with student data
student_names = ["Mrunmayi", "Rahul", "Priya", "Aman"]
student_marks = (85, 78, 92, 85)
unique_marks = {85, 78, 92}
student_details = {
    "name": "Mrunmayi",
    "branch": "CSE-DS",
    "marks": 85
}
print("List:", student_names)
print("Tuple:", student_marks)
print("Set:", unique_marks)
print("Dictionary:", student_details)
```

```python
#Comparison of Python collections
collections = {
    "List": "Ordered, Changeable, Allows duplicates",
    "Tuple": "Ordered, Unchangeable, Allows duplicates",
    "Set": "Unordered, Changeable, No duplicates",
    "Dictionary": "Key-Value pairs, Changeable, Keys are unique"
}
for collection, features in collections.items():
    print(collection, ":", features)
```
